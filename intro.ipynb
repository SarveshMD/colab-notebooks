```python
M1 = torch.tensor([[1,2],[3,4]])
M2 = torch.tensor([[4,3],[7,9]])
```

```python
def run():
  matrix_mux = 0
  res = torch.zeros((M1.shape[0], M2.shape[1]))
  for i in range(M1.shape[0]):
    for j in range(M2.shape[1]):
      res[i][j] = torch.dot(M1[i], M2[:, j])
%timeit run()
```
